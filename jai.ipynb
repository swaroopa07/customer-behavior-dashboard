```
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median))
```

```
df.isnull().sum()
```

```
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```
df.columns
```

```
# age groups by age
label = ['Young Adult', 'Adult' , 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(df['age'], q=4, labels=label)
```

```
df[['age','age_group']].head(10)
```

```
# column purchase_frequent_days

frequent_days = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months' : 90
}

df['purchase_frequent_days'] = df['frequency_of_purchases'].map(frequent_days)
```

```
df[['frequency_of_purchases', 'purchase_frequent_days']].head(5)
```

```
df['review_rating'] = pd.to_numeric(df['review_rating'], errors='coerce')
```
